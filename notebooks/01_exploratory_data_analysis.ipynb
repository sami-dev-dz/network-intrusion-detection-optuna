```python
import pandas as pd
import numpy as np

path = "../data/UNSW_NB15_training-set.parquet"

# Chargement du fichier Parquet
df = pd.read_parquet(path)
df.head()
```

```python
# 2. Taille et structure globale
print("=== 1. DIMENSIONS DU DATASET ===")
print(f"Nombre d'observations (lignes) : {df.shape[0]:,}")
print(f"Nombre de variables (colonnes) : {df.shape[1]}")
```

```python
# 3. Types de variables
print("\n=== 2. TYPES DE VARIABLES ===")
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
num_cols = df.select_dtypes(include=['number']).columns.tolist()
print(f"Variables numériques ({len(num_cols)})")
print(f"Variables catégorielles ({len(cat_cols)})")
```

```python
# 4. Distribution de la variable cible (Classes)
print("\n=== 3. DISTRIBUTION DES CLASSES (label) ===")
counts = df['label'].value_counts()
percents = (df['label'].value_counts(normalize=True) * 100).round(2)
class_dist = pd.DataFrame({
    'Effectif': [counts[0], counts[1]],
    'Pourcentage (%)': [percents[0], percents[1]]
}, index=['Normal (0)', 'Attaque (1)'])

display(class_dist)
```

```python
# Distribution détaillée par famille d'attaque
if 'attack_cat' in df.columns:
    print("\n--- Détail par catégorie d'attaque ---")
    attack_dist = pd.DataFrame({
        'Effectif': df['attack_cat'].value_counts(),
        'Pourcentage (%)': (df['attack_cat'].value_counts(normalize=True) * 100).round(2)
    })
    display(attack_dist)
```

```python
# 5. Valeurs manquantes
print("\n=== 4. INTÉGRITÉ DES DONNÉES (VALEURS MANQUANTES) ===")
missing = df.isnull().sum()
missing_cols = missing[missing > 0]

if len(missing_cols) == 0:
    print("Aucune valeur manquante (NaN) détectée dans l'ensemble du dataset.")
else:
    print("Colonnes avec valeurs manquantes :")
    print(missing_cols)
```

```python
# Vérification bonus des valeurs infinies (fréquentes dans les données réseau)
inf_count = np.isinf(df[num_cols]).sum().sum()
print(f"Total des valeurs infinies (inf) : {inf_count}")
```
